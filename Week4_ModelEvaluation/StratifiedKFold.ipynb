### Import Packages

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

from sklearn.linear_model import LogisticRegression
```

### Load data:

```python
X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y)
```

### Create pipeline:

```python
pipe = make_pipeline(StandardScaler(), PCA(n_components=2), LogisticRegression(solver='lbfgs'))
pipe.fit(X_train, y_train)
y_pred = pipe.predict(X_test)
pipe.score(X_test, y_test)
```

## cross_val_score:

```python
cross_val_score(pipe,
                X_train, 
                y_train,
                cv=10 # split data to 10 folds
                )
```

## StratifiedKFold:
- StratifiedKFold(n_splits=10).split(X_train, y_train)
    - n_splits
    - split(X_train, y_train)
- StratifiedKFold would generate the training index and test index for each split
- Then use the .score() to generate the score for each split

```python
kfold = StratifiedKFold(n_splits=10).split(X_train, y_train)
```

```python
scores = []
for i, (train_idx, test_idx) in enumerate(kfold):
    pipe.fit(X_train[train_idx], y_train[train_idx])
    score = pipe.score(X_train[test_idx], y_train[test_idx])
    scores.append(score)
    print(f'Fold: {i+1}, accuracy: {score:.4f}')
```

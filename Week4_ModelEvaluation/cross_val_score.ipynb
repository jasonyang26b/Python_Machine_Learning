## **cross_val_score:**

`cross_val_score` is a function in `scikit-learn` that evaluates the performance of a machine learning model by using **k-fold cross-validation**. It splits the dataset into $k$ subsets (folds), trains the model on $k-1$ folds, and validates it on the remaining fold, repeating this process $k$ times.

### Purpose
It provides a more robust estimate of a model's performance on unseen data compared to a single train/test split, as it ensures every data point is used for both training and validation.

### Syntax
```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(estimator, X, y, scoring=None, cv=None, n_jobs=None)
```

### Parameters
*   **`estimator`**: The machine learning model (e.g., `RandomForestClassifier`, `SVC`) that implements the `.fit()` and `.predict()` methods.
*   **`X`**: The feature matrix (independent variables).
*   **`y`**: The target vector (dependent variable).
*   **`scoring`**: 
    *   `None` (default): Uses the estimator's default `.score()` method.
    *   `str`: A predefined string identifier (e.g., `'accuracy'`, `'f1'`, `'neg_mean_squared_error'`).
    *   `callable`: A scorer object created via `make_scorer`.
*   **`cv`**: 
    *   `None` (default): Uses 5-fold cross-validation.
    *   `int`: Specifies the number of folds (e.g., `cv=10`).
    *   `CrossValidator` object: A specific splitting strategy (e.g., `StratifiedKFold`).
*   **`n_jobs`**: The number of CPU cores to use for parallel computation. `-1` uses all available processors.

### Example Usage

```python
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris

# Load data
data = load_iris()
X, y = data.data, data.target

# Initialize model
clf = RandomForestClassifier(n_estimators=100)

# Perform 5-fold cross-validation using accuracy
scores = cross_val_score(clf, X, y, cv=5, scoring='accuracy')

print(f"Scores per fold: {scores}")
print(f"Mean Accuracy: {scores.mean():.2f}")
print(f"Standard Deviation: {scores.std():.2f}")
```

### Key Characteristics
1.  **Returns an Array**: It returns an array of scores (one for each fold), not a single number. To get a single metric, you typically take the `.mean()`.
2.  **No Model Training Persistence**: `cross_val_score` does **not** return a trained model. It is used strictly for evaluation. If you need the model itself after cross-validation, use `cross_validate` (which returns training/test scores) or fit the model on the full dataset manually.
3.  **Stratification**: When using classification with integer `cv`, scikit-learn uses `StratifiedKFold` by default, ensuring each fold has the same percentage of samples of each class as the complete set.
#
---

The scores returned by `cross_val_score` are calculated based on **validation data** (often called "hold-out" or "test" data within the context of each fold), not the training data used for that specific iteration.

Here is the breakdown of how it works:

### 1. The Process per Fold
In each iteration of $k$-fold cross-validation:
1.  **Split:** The data is split into two parts: **Training Set** and **Validation Set**.
2.  **Train:** The model is trained using only the **Training Set**.
3.  **Predict:** The trained model makes predictions on the **Validation Set**.
4.  **Score:** The metric (e.g., accuracy) is calculated by comparing those predictions against the actual labels of the **Validation Set**.

### 2. Why it is not "Testing Data"
In a strict machine learning workflow, there are three distinct datasets:
*   **Training Set:** Used to teach the model.
*   **Validation Set:** Used to evaluate the model during tuning (this is what `cross_val_score` uses).
*   **Test Set:** A completely unseen dataset used **only once** at the very end to report the final performance.

**Crucial Rule:** You should never use `cross_val_score` on your final Test Set. You use it on your Training/Validation data to find the best model or parameters. Once you have chosen your best model, you perform a single final evaluation on the Test Set.

### Summary Table
| Dataset | Used by `cross_val_score`? | Purpose |
| :--- | :--- | :--- |
| **Training Fold** | Yes (for `.fit()`) | To learn patterns in the data. |
| **Validation Fold** | **Yes (for `.score()`)** | To evaluate how well the model generalizes. |
| **Final Test Set** | **No** | To provide an unbiased final performance report. |

## Example:

````python
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris

# load data
data = load_iris()
X, y = data.data, data.target

# Initialize model
clf = RandomForestClassifier(n_estimators=100)

# Perform 5-fold cross_validation using accuracy
scores = cross_val_score(clf, X, y, cv=5, scoring='accuracy')

print(f'Scores per fold: {scores}')
print(f'Mean Accuracy: {scores.mean():.2f}')
print(f'Standard Deviation: {scores.std():.2f}')
````
